```python
# 函数 fucntion
def add(x, y):
    return x + y
result = add(3, 5)
print("函数 add 的结果:", result)  # 输出: 函数 add 的结果: 8
```

```python
# lambda（匿名函数）
square = lambda x: x ** 2
result = square(4)
print("lambda 函数的结果:", result)  # 输出: lambda 函数的结果: 16
```

```python
# map（映射函数）
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x ** 2, numbers))
print("map 函数的结果:", squared_numbers)  # 输出: map 函数的结果: [1, 4, 9, 16, 25]
```

```python
# filter（过滤函数）
numbers = [1, 2, 3, 4, 5]
filtered_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print("filter 函数的结果:", filtered_numbers)  # 输出: filter 函数的结果: [2, 4]
```

```python
# reduce（归约函数）
from functools import reduce
numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
print("reduce 函数的结果:", product)  # 输出: reduce 函数的结果: 120


```

# reduce（归约函数） 代码在做什么
'''
reduce(lambda x, y: x * y, numbers) 的含义是：从左到右，把列表里的数两两交给 lambda x, y: x * y，每次的结果再和下一个数继续乘，直到只剩一个值。

等价于手动写：

((((1 * 2) * 3) * 4) * 5)
逐步执行（内部“折叠”过程）
numbers = [1, 2, 3, 4, 5] 时，大致等价于下面这种循环逻辑（便于理解，不是 C 源码）：

取前两个元素：x=1, y=2 → 1 * 2 = 2
把上一步结果当作新的 x，下一个元素当 y：x=2, y=3 → 2 * 3 = 6
x=6, y=4 → 6 * 4 = 24
x=24, y=5 → 24 * 5 = 120
所以最终 product == 120。

和“普通循环”的对应关系
可以把它想成：

acc = numbers[0]
for y in numbers[1:]:
    acc = acc * y  # 这里就是 lambda(acc, y)
reduce 只是把“累加器 acc + 二元函数”这种写法抽象成了一个高阶函数。
'''
小结
第一次：函数收到的是序列里前两个元素。
之后每一次：第一个参数是上一次的返回值，第二个参数是序列里的下一个元素。
结束：序列遍历完，最后一次调用的返回值就是 reduce 的结果。
若你传入第三个参数 initial，第一次会变成 func(initial, numbers[0])，再依次和后面元素组合；你当前代码没有 initial，所以从 numbers[0] 和 numbers[1] 开始。

```python
# zip （压缩函数）
# 并行聚合多个可迭代的元素，长度以最短为准（返回迭代器）
names = ["Alice", "Bob", "Charlie"]
ages = [30, 25, 35]
zipped = list(zip(names, ages))
print("zip 函数的结果:", zipped)  # 输出: zip 函数的结果: [('Alice', 30), ('Bob', 25), ('Charlie', 35)]
```

```python
# enumerate（枚举函数）
# 同时获取元素的索引和值
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(f"索引: {index}, 水果: {fruit}")
# 输出:
# 索引: 0, 水果: apple
# 索引: 1, 水果: banana
# 索引: 2, 水果: cherry
```

```python
# sorted（排序函数）
# 默认升序
numbers = [5, 2, 9, 1, 5, 6]
sorted_numbers = sorted(numbers)
print("sorted 函数的结果:", sorted_numbers)  # 输出: sorted 函数的结果: [1, 2, 5, 5, 6, 9]

# 按照指定排序 （例如按绝对值排序）
numbers = [-5, -2, 9, 1, -5, 6]
sorted_numbers = sorted(numbers, key=abs)
print("按绝对值排序的结果:", sorted_numbers)  # 输出: 按绝对值排序的结果: [1, -2, 6, -5, -5, 9]

# reverse 参数
numbers = [5, 2, 9, 1, 5, 6]
sorted_numbers_desc = sorted(numbers, reverse=True)
print("降序排序的结果:", sorted_numbers_desc)  # 输出: 降序排序的结果: [9, 6, 5, 5, 2, 1]
```

```python
# reversed（反转函数）
# 返回一个反转的迭代器
numbers = [1, 2, 3, 4, 5]
reversed_numbers = list(reversed(numbers))
print("reversed 函数的结果:", reversed_numbers)  # 输出: reversed  函数的结果: [5, 4, 3, 2, 1]
```

```python
# any（任意函数）
# 只要有一个元素为 True 就返回 True
bools = [False, False, True, False]
result = any(bools)
print("any 函数的结果:", result)  # 输出: any 函数的结果: True
```

```python
# list comprehension（列表推导式）
# 生成一个包含 0-9 的平方的列表
squares = [x ** 2 for x in range(10)]
print("列表推导式的结果:", squares)  # 输出: 列表推导式的结果: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
```

```python
# list()（列表函数）
# 将字符串转换为列表
s = "hello"
char_list = list(s)
print("list() 函数的结果:", char_list)  # 输出: list() 函数的结果: ['h', 'e', 'l', 'l', 'o
```
