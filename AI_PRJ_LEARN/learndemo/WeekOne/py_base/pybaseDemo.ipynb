```python
# dict（字典）：无序、可修改、键唯一
person = {"name": "Alice", "age": 30}
person["city"] = "New York"  # 添加键值对
print("dict:",person)     # 输出: {'name': 'Alice', 'age': 30, 'city': 'New York'}
 # 获取值
print("person name:",person.get("name") )
```

```python
# set（集合）：无序、可修改、元素唯一
unique_nums = {11,1, 2, 3, 4, 5}
unique_nums.add(6)  # 添加元素
print("set:",unique_nums)   # 输出: {1, 2, 3, 4, 5, 6}
unique_nums.add(3)  # 添加重复元素不会改变集合
print("set after adding duplicate:",unique_nums)  # 输出: {1, 2, 3, 4, 5, 6}
unique_nums.add(0)      # 添加新元素
print("set after adding new element:",unique_nums)  # 输出: {0, 1, 2, 3, 4, 5, 6}
unique_nums.remove(2)   # 删除元素
print("set after removing element:",unique_nums)  # 输出: {0, 1
unique_nums.add(4)
print(unique_nums)
```

```python
# stack（栈）和 queue（队列）是两种常见的数据结构，分别遵循不同的访问原则。
# 1) stack（栈）：后进先出（LIFO）
stack = []
stack.append(1)  # 入栈
stack.append(2)
stack.append(3)
print("stack:",stack)  # 输出: [1, 2, 3]
top_element = stack.pop()  # 出栈
print("popped element:",top_element)  # 输出: 3
print("stack after pop:",stack)  # 输出: [1, 2]
```

```python
# 2) queue（队列）：先进先出（FIFO）
from collections import deque
queue = deque()
queue.append(1)  # 入队
queue.append(2)
queue.append(3)
print("queue:",queue)  # 输出: deque([1, 2, 3])
front_element = queue.popleft()  # 出队
print("dequeued element:",front_element)  # 输出: 1
print("queue after dequeue:",queue)  # 输出: deque([2, 3])
```

```python
# 3) deque（双端队列）：两端都可以添加和删除元素
deque_example = deque()
deque_example.append(1)  # 从右端添加
deque_example.appendleft(0)  # 从左端添加
print("deque:",deque_example)  # 输出: deque([0, 1])
right_element = deque_example.pop()  # 从右端删除
print("popped from right:",right_element)  # 输出: 1
print("deque after popping from right:",deque_example)  # 输出: deque([0])
left_element = deque_example.popleft()  # 从左端删除
print("popped from left:",left_element)  # 输出: 0
print("deque after popping from left:",deque_example)  # 输出: deque([])
```

```python
# 4) heap（堆）：优先级队列，最小堆或最大堆
import heapq
min_heap = []
heapq.heappush(min_heap, 5)  # 添加元素
heapq.heappush(min_heap, 3)
heapq.heappush(min_heap, 8)
print("min_heap:",min_heap)  # 输出: [3, 5, 8]
smallest = heapq.heappop(min_heap)  # 弹出最小元素
print("smallest element:",smallest)  # 输出: 3
print("min_heap after pop:",min_heap)  # 输出: [5, 8]
```

```python
# 异常try except finally
try:
    print("try")
    raise ValueError("这是一个错误")  # 主动抛出一个异常
except ValueError as e:
    print("except:", e)  # 捕获并打印异常信息
finally:
    print("finally")  # 无论是否发生异常都会执行
```
