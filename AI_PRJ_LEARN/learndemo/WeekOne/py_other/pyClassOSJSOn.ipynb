```python
# 迭代器和生成器
# 迭代器
my_list = [1, 2, 3]
my_iter = iter(my_list)  # 创建一个迭代器对象
print("迭代器:", next(my_iter))  # 输出: 1
print("迭代器:", next(my_iter))  # 输出: 2
print("迭代器:", next(my_iter))  # 输出: 3
# next(my_iter)  # 再次调用会引发 StopIteration 异常
```

```python
# 生成器
def count_up_to(n):
    """这是一个生成器函数，生成从 1 到 n 的数字"""
    count = 1
    while count <= n:
        yield count  # 使用 yield 生成一个值
        count += 1  # 继续生成下一个值
counter = count_up_to(5)  # 创建一个生成器对象
print("生成器:", next(counter))  # 输出: 1
print("生成器:", next(counter))  # 输出: 2
print("生成器:", next(counter))  # 输出: 3
print("生成器:", next(counter))  # 输出: 4

```

```python
# 生成器 进阶
def countdown(n):
    while n > 0:
        yield n
        n -= 1

# 创建生成器对象
generator = countdown(5)

# 通过迭代生成器获取值
print(next(generator))  # 输出: 5
print(next(generator))  # 输出: 4
print(next(generator))  # 输出: 3

# 使用 for 循环迭代生成器
for value in generator:
    print(value)  # 输出: 2 1
```

```python
# 上下文管理器
class MyContextManager:
    def __enter__(self):
        print("进入上下文管理器")  # 进入上下文时执行的代码
        return self  # 可以返回一个对象供 with 块使用
    def __exit__(self, exc_type, exc_value, traceback):
        print("退出上下文管理器")  # 退出上下文时执行的代码
        if exc_type:
            print(f"发生异常: {exc_value}")  # 如果发生异常，打印异常信息
        return True  # 返回 True 表示异常已处理，不会传播
# 使用上下文管理器
with MyContextManager() as manager:
    print("在 with 块内")  # 在 with 块内执行的代码
    # raise ValueError("这是一个错误")  # 可以在 with 块内主动抛出一个异常来测试 __exit__ 方法的异常处理
```

```python
'''
最佳实践
优先使用 with 管理资源：对于文件、网络连接、锁等资源，总是优先考虑使用 with 语句
保持上下文简洁：with 块中的代码应该只包含与资源相关的操作
合理处理异常：在自定义上下文管理器中，根据需求决定是否抑制异常
利用多个上下文：Python 允许在单个 with 语句中管理多个资源
'''
# with 语句（上下文管理器）
# 自动管理资源（例如文件）
with open("../file/example.txt", "w") as f:
    f.write("Hello, world!")  # 在 with 块内写入内容
    f.write("\nThis is a file example.")  # 可以继续写入多行内容
# 文件会在 with 块结束时自动关闭，无需手动调用 f.close()
```

```python
from contextlib import contextmanager

@contextmanager
def my_resource():
    print("资源获取")
    try:
        yield "资源对象"
    finally:
        print("资源释放")

with my_resource() as res:
    print(f"使用: {res}")
```

```python
# 函数定义和调用
def greet(name):
    """这是一个简单的问候函数"""
    return f"Hello, {name}!"
```

```python
# 调用函数
greeting = greet("Alice")
print(greeting)  # 输出: Hello, Alice!
```

```python
# 文件读取 和写入
# 写入文件
with open("../file/example.txt", "w") as f:
    f.write("Hello, world!\nThis is a file example.")  # 写入多行内容
# 读取文件
with open("../file/example.txt", "r") as f:
    content = f.read()  # 读取整个文件内容
    print("File content:\n", content)  # 输出文件内容
```

```python
# 类定义和实例化
class Person:
    # 类有一个名为 __init__() 的特殊方法（构造方法）
    def __init__(self, name, age):
        self.name = name  # 实例属性
        self.age = age  # 实例属性
    # 类方法（实例方法）
    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."  # 实例方法
# 创建类的实例
person1 = Person("Alice", 30)
# 调用实例方法
greeting = person1.greet()
print(greeting)  # 输出: Hello, my name is Alice and I am 30 years old.
```

```python
# 装饰器 （函数装饰器）
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("装饰器开始执行")  # 在调用原函数前执行的代码
        result = func(*args, **kwargs)  # 调用原函数
        print("装饰器结束执行")  # 在调用原函数后执行的代码
        return result  # 返回原函数的结果
    return wrapper
# 使用装饰器
@my_decorator
def say_hello(name):
    print(f"Hello, {name}!")  # 原函数的功能
# 调用被装饰的函数
say_hello("Alice")  # 输出: 装饰器开始执行 Hello, Alice! 装饰器结束执行
```

```python
# Json 模块
import json
# Python 对象转 JSON 字符串
data = {"name": "Alice", "age": 30, "city": "New York"}
json_str = json.dumps(data)  # 将 Python 对象转换为 JSON 字符串
print("JSON 字符串:", json_str)  # 输出: JSON 字符串: {"name": "Alice", "age": 30, "city": "New York"}
# JSON 字符串转 Python 对象
json_str = '{"name": "Alice", "age": 30, "city": "  New York"}'
data = json.loads(json_str)  # 将 JSON 字符串转换为 Python 对象
print("Python 对象:", data)  # 输出: Python 对象: {'name': 'Alice', 'age': 30, 'city': 'New York'}
```
