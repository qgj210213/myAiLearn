```python
# 迭代器和生成器
# 迭代器
my_list = [1, 2, 3]
my_iter = iter(my_list)  # 创建一个迭代器对象
print("迭代器:", next(my_iter))  # 输出: 1
print("迭代器:", next(my_iter))  # 输出: 2
print("迭代器:", next(my_iter))  # 输出: 3
# next(my_iter)  # 再次调用会引发 StopIteration 异常

# 生成器
def count_up_to(n):
    """这是一个生成器函数，生成从 1 到 n 的数字"""
    count = 1
    while count <= n:
        yield count  # 使用 yield 生成一个值
        count += 1  # 继续生成下一个值
counter = count_up_to(5)  # 创建一个生成器对象
print("生成器:", next(counter))  # 输出: 1
print("生成器:", next(counter))  # 输出: 2
print("生成器:", next(counter))  # 输出: 3
print("生成器:", next(counter))  # 输出: 4
```

```python

# 异常try except finally
try:
    print("try")
    raise ValueError("这是一个错误")  # 主动抛出一个异常
except ValueError as e:
    print("except:", e)  # 捕获并打印异常信息
finally:
    print("finally")  # 无论是否发生异常都会执行
```

```python
# with 语句（上下文管理器）
# 自动管理资源（例如文件）
with open("../file/example.txt", "w") as f:
    f.write("Hello, world!")  # 在 with 块内写入内容
    f.write("\nThis is a file example.")  # 可以继续写入多行内容
# 文件会在 with 块结束时自动关闭，无需手动调用 f.close()
```

```python
# 函数定义和调用
def greet(name):
    """这是一个简单的问候函数"""
    return f"Hello, {name}!"
```

```python
# 调用函数
greeting = greet("Alice")
print(greeting)  # 输出: Hello, Alice!
```

```python
# 文件读取 和写入
# 写入文件
with open("../file/example.txt", "w") as f:
    f.write("Hello, world!\nThis is a file example.")  # 写入多行内容
# 读取文件
with open("../file/example.txt", "r") as f:
    content = f.read()  # 读取整个文件内容
    print("File content:\n", content)  # 输出文件内容
```

```python
# 类定义和实例化
class Person:
    # 类有一个名为 __init__() 的特殊方法（构造方法）
    def __init__(self, name, age):
        self.name = name  # 实例属性
        self.age = age  # 实例属性
    # 类方法（实例方法）
    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."  # 实例方法
# 创建类的实例
person1 = Person("Alice", 30)
# 调用实例方法
greeting = person1.greet()
print(greeting)  # 输出: Hello, my name is Alice and I am 30 years old.
```
